# Completely Randomized Design

## Example: Drug Dosage and Memory

In a clinical experiment, researchers want to test how different dosages of a new medication affect memory performance. Participants are randomly assigned to one of three groups: one receives a placebo (0 mg), another receives a low dose (50 mg), and the last group receives a high dose (100 mg). After taking the assigned treatment, all participants complete a standardized memory test.

### Treatment Factor

- Drug dosage

This is the independent variable that the experimenter manipulates.

### Treatments

- Placebo (0 mg)
- Low dose (50 mg)
- High dose (100 mg)

These are the specific conditions applied to participants. If you’re confused by the term treatments, it may help to think of them as treatment groups instead.

### Experimental Unit

- Individual participant

It’s the smallest subject or item that can receive a different treatment from others.

### Completely randomized design

In a completely randomized design, all experimental units are randomly assigned to treatments, and each unit receives only one treatment.

# Completely Randomized Block Design

## Example: Drug Dosage and Memory

As an extension of the previous drug study, researchers now suspect that sex may influence memory performance. To account for this, participants are first blocked by sex (male, female), and treatments (placebo, low dose, high dose) are randomly assigned within each block. This design reduces variability due to sex and ensures balanced treatment comparisons across blocks.

### Treatment Factor & Treatments

- Same as the above

### Blocking Factor

- Sex

Note: In this study, our primary interest is in the effect of drug dosage, not in the effect of sex itself. Think of yourself as a researcher hired by a pharmaceutical company - would you be studying the effect of sex? Probably not. Your focus would be on how different dosages affect outcomes. Sex is considered a blocking factor - a variable that may influence the treatment effect, and by accounting for it through blocking, we reduce unwanted variation and increase the precision of our comparisons.

### Experimental Unit

- Same as the above

### Completely Randomized Block Design

In a completely randomized block design, experimental units are first grouped into blocks based on a variable that may affect the outcome. Then, within each block, units are randomly assigned to treatments.
